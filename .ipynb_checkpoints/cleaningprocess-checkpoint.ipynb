```python
import pandas as pd

jobs = pd.read_excel("data/indian-job-market-dataset-2025.xlsx")
print("Shape:", jobs.shape)
jobs.head(3)
```

```python
clean = jobs.dropna(subset=["tagsAndSkills", "minimumExperience", "maximumExperience"])

clean["skills_list"] = clean["tagsAndSkills"].apply(lambda x: [s.strip() for s in x.split(",")])

clean["minimumSalary"] = clean["minimumSalary"].replace(0, None)
clean["maximumSalary"] = clean["maximumSalary"].replace(0, None)

print("Before cleaning:", jobs.shape[0], "rows")
print("After cleaning:", clean.shape[0], "rows")
clean[["title", "skills_list", "minimumExperience", "maximumExperience"]].head(5)
```

```python
clean.to_csv("data/cleaned_jobs.csv", index=False)
print("Saved cleaned dataset!")
```
